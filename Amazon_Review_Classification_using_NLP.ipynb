```python
# ===============================================================
# Project Name : Amazon Review Classification using NLP
# Algorithm    : Multinomial Naive Bayes

# ===============================================================

# ===============================================================
# Step 1 : Import Required Libraries
# ===============================================================

import pandas as pd
import joblib

# Used for splitting dataset

from sklearn.model_selection import train_test_split

# Used for converting text into numerical vectors

from sklearn.feature_extraction.text import TfidfVectorizer

# Machine Learning Algorithm

from sklearn.naive_bayes import MultinomialNB

# Used for model evaluation

from sklearn.metrics import accuracy_score, classification_report

# ===============================================================
# Step 2 : Create Amazon Review Dataset
# ===============================================================

data = {

"Review":[

# ===========================
# Positive Reviews
# ===========================

"This product is amazing",

"Excellent quality and fast delivery",

"Highly recommended for everyone",

"Battery backup is outstanding",

"The mobile phone works perfectly",

"Very comfortable shoes",

"The laptop performance is excellent",

"Good packaging and original product",

"Five stars for this amazing item",

"Worth every penny",

# ===========================
# Negative Reviews
# ===========================

"Very poor quality",

"Completely waste of money",

"Battery drains very fast",

"Received damaged product",

"The charger stopped working",

"Very disappointed with the purchase",

"Bad customer support",

"The product is fake",

"Screen quality is terrible",

"I will never buy this again"

],

"Sentiment":[

# Positive

"Positive",
"Positive",
"Positive",
"Positive",
"Positive",
"Positive",
"Positive",
"Positive",
"Positive",
"Positive",

# Negative

"Negative",
"Negative",
"Negative",
"Negative",
"Negative",
"Negative",
"Negative",
"Negative",
"Negative",
"Negative"

]

}

# ===============================================================
# Step 3 : Convert Dictionary into DataFrame
# ===============================================================

df = pd.DataFrame(data)

print("========================================")

print("Amazon Review Dataset")

print("========================================")

print(df)

# ===============================================================
# Step 4 : Separate Features and Target
# ===============================================================

X = df["Review"]

# Input Reviews

y = df["Sentiment"]

# Output Labels

# ===============================================================
# Step 5 : Convert Text into Numerical Features
# ===============================================================

vectorizer = TfidfVectorizer(

stop_words="english",

lowercase=True

)

X_vector = vectorizer.fit_transform(X)

print("\nVocabulary Size :",len(vectorizer.get_feature_names_out()))

# ===============================================================
# Step 6 : Split Dataset
# ===============================================================

X_train,X_test,y_train,y_test = train_test_split(

X_vector,

y,

test_size=0.20,

random_state=42

)

print("\nTraining Samples :",X_train.shape[0])

print("Testing Samples :",X_test.shape[0])

# ===============================================================
# Step 7 : Create Machine Learning Model
# ===============================================================

model = MultinomialNB()

# ===============================================================
# Step 8 : Train Model
# ===============================================================

model.fit(

X_train,

y_train

)

print("\nModel Training Completed Successfully")

# ===============================================================
# Step 9 : Prediction
# ===============================================================

prediction = model.predict(

X_test

)

# ===============================================================
# Step 10 : Evaluate Model
# ===============================================================

accuracy = accuracy_score(

y_test,

prediction

)

print("\n========================================")

print("Model Accuracy")

print("========================================")

print(accuracy)

print("\nClassification Report")

print(classification_report(

y_test,

prediction

))

# ===============================================================
# Step 11 : Predict User Review
# ===============================================================

review = [

input("\nEnter Amazon Review : ")

]

review_vector = vectorizer.transform(

review

)

result = model.predict(

review_vector

)

print("\nPredicted Sentiment :",result[0])

# ===============================================================
# Step 12 : Save Model
# ===============================================================

joblib.dump(

model,

"model.pkl"

)

joblib.dump(

vectorizer,

"vectorizer.pkl"

)

print("\nModel Saved Successfully")

print("model.pkl")

print("vectorizer.pkl")

# ===============================================================
# Step 13 : Display Vocabulary Size
# ===============================================================

print("\nVocabulary Size :")

print(len(vectorizer.vocabulary_))

# ===============================================================
# Step 14 : Display Sample Features
# ===============================================================

print("\nSample Features :")

print(vectorizer.get_feature_names_out()[:30])

# ===============================================================
# Step 15 : Project Completed
# ===============================================================

print("\n========================================")

print("Amazon Review Classification Model is Ready")

print("========================================")
```
